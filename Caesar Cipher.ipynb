As part of the workshop material, you have already worked through the basic concepts of the Caesar cipher, specifically how to encrypt and decrypt using it. We have some code that looks like the following (which you can use without alteration - just run the cell).

````python
def caesar_cipher_key(shift=3,punctuation=". ?,!"):
    """Produce the key for the Caesar Cipher
    Inputs: shift - the offset which we use of one alphabet relative to the other. Default: 3
            punctuation - the allowed punctuation, which would not be translated.
    Output: the key"""
    assert isinstance(shift,int)
    assert 0<=shift<26
    assert isinstance(punctuation,str)
    alphabet="ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    key=alphabet[shift:]+alphabet[:shift]+punctuation
    assert len(key)==26+len(punctuation)
    return key

def encrypt(plain_text,cipher):
    """This function will encrypt the plain_text using the key cipher"""
    alphabet="ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    #preconditions
    assert isinstance(cipher,str)
    assert len(cipher)>=26
    alphabet+=cipher[26:]                                      # add the punctuation to our alphabet
    assert all([letter in alphabet for letter in cipher[:26]]) # first 26 characters are letters in the alphabet
    assert len(set(cipher[:26]))==26                           # make sure each of the first 26 letters is unique, no repetitions
    assert all(letter in alphabet for letter in plain_text)     # make sure that all the characters in the plain_text are in the alphabet
    
    #the following is the main computational step. Just the inverse of encryption
    output="".join([cipher[alphabet.index(letter)] for letter in plain_text])
    
    #post conditions
    assert len(plain_text)==len(output)
    return output

def decrypt(cipher_text,cipher):
    """This function will decrypt the encrypted text cipher_text using the key cipher"""
    alphabet="ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    #preconditions
    assert isinstance(cipher,str)
    assert len(cipher)>=26
    assert all([letter in alphabet for letter in cipher[:26]]) # first 26 characters are letters in the alphabet
    assert len(set(cipher[:26]))==26                           # make sure each of the first 26 letters is unique, no repetitions
    assert all(letter in cipher for letter in cipher_text),set([letter for letter in cipher_text if letter not in cipher])     # make sure that all the characters in the cipher_text are in the key
    
    alphabet+=cipher[26:]                                      # add the punctuation to our alphabet
    
    #the following is the main computational step. Just the inverse of encryption
    output="".join([alphabet[cipher.index(letter)] for letter in cipher_text])
    
    #post conditions
    assert len(cipher_text)==len(output)
    return output
````

## Vector Inner Product

If we compute `get_freqs` for a given piece of text, that's basically a probability distribution for the letters through the text. If we compare two probability distributions, we can see how similar they are. How to make that comparison? Well, we can also think of the output of `get_freqs` as a vector. How do we compare how similar two vectors are? Measure the angle between them!

Recall that if you have two vectors $\underline v_1$ and $\underline v_2$, you can calculate their inner product,
$$\underline v_1\cdot\underline v_2=\|\underline v_1\|\|\underline v_2\|\cos\theta=\sum_{i=1}^N (\underline{v}_1)_i(\underline{v}_2)_i,$$
where $\theta$ is the angle between the two vectors, measuring how similar the directions of the two vectors are.

**Complete the function `inner_product`** that accepts $\underline v_1$ and $\underline v_2$ as lists, and returns their inner product. You must code this yourself and not rely on any external functions.

Don't forget pre-conditions!

<div style='text-align: right; font-style:italic; font-size: large;'>Marks: 5</div>

````python
def inner_product(list1, list2):
    '''input: two lists of real numbers of the same length
    output: the inner product of the two lists'''

    #preconditions
    assert isinstance(list1, list)
    assert isinstance(list2, list)
    assert len(list1) == len(list2)

    return sum([list1[i] * list2[i] for i in range(len(list1))])
````

The `cycle_list` operation is defined as taking a list (first argument), removing one element from the front, and putting it back on the end. This can be repeated k times (optional second argument). **Complete the `cycle_list` function.**

*Hint*: Recall the strong recommendations you've been given for function creation and list manipulation:
- if you need to copy a list, use `new_list=list(old_list)`, NOT `new_list=old_list`.
- if one of your function arguments is `old_list`, do not change `old_list` within your function. Create a copy and only alter the copy.
- never call a list `list`

<div style='text-align: right; font-style:italic; font-size: large;'>Marks: 2</div>

````python
def cycle_list(myList, k=1):
    """
    input: a list and a cycle length
    output: the list cycled k times (defaults to 1)
    """
    new_list = list(myList)

    for i in range(k):
        first = new_list.pop(0)
        new_list.append(first)

    return new_list
````

For each of the 26 different possible cycle lengths, we want to apply that cycle to the frequency information of the encrypted text, and calculate the inner product with the reference frequency information. **Complete the `all_cycles` function.** How this works is defined in the docstring. Once we have the data from `all_cycles`, `get_best_cycle_length` will extract the one with the largest inner product (i.e. smallest angle between the vectors). This part of the code is already done for you. The expectation is that this should be the best choice for offset.

<div style='text-align: right; font-style:italic; font-size: large;'>Marks: 4</div>

````python
def get_best_cycle_length():
    '''
    Inputs: none
    Outputs: length of cycle that gives largest inner product.
    You do not need to change this function
    '''
    #test all the possible Caesar cipher offsets
    ans=all_cycles(encrypt_freqs,ref_freqs)
    return ans.index(max(ans))    #find the largest entry in the list, and return its position in the list. that's the desired offset

def all_cycles(list1,list2):
    '''
    Input: two lists of equal length containing frequency data (e.g. ouptput from get_freqs for two different sources)
    Output: list that is the same length as the input. Element n (0-indexed) is inner product between n-cycle of first list, and second list'''
    results = []
    for n in range(len(list1)):
        results.append(inner_product(cycle_list(list1, n), list2))
    return results
    raise NotImplementedError()
````
