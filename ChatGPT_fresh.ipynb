One problem that we run into when importing data from other sources is that it's often in very inconsistent formats, and we have to convert them into something more regular in order to be able to extract useful information. One example of this is white space (spaces) being scatted around. We want to remove it.

Get a free version of [ChatGPT](https://chatgpt.com/) to solve the following (you can copy & paste it as the prompt):

Using only the Python standard library (no third-party packages such as nltk or spaCy), and providing only code and example usage without further commentary, please write a python function called `clean_string` that takes a string and does the following:
- removes any white space from the beginning and end of the string
- if there are multiple white spaces in a row, replace them with just a single one
- make sure the beginning of each sentence is a capital letter

<br /><br /><br />

Enter the given function below, including example usage.
If the function doesn't make any sense to you (it may have chosen to use a lot of functions that you've never met before), ask it to give you code without using any imports.

````python
#copy the function that chatgpt creates here, including example usage.
# YOUR CODE HERE
#ChatGPT prompted by Pranav
def clean_string(text: str) -> str:
    if not isinstance(text, str):
        raise TypeError("text must be a string")

    # Remove leading and trailing whitespace
    text = text.strip()

    # Replace multiple whitespace characters with a single space
    text = " ".join(text.split())

    # Capitalize the beginning of each sentence
    result = []
    capitalize_next = True

    for char in text:
        if capitalize_next and char.isalpha():
            result.append(char.upper())
            capitalize_next = False
        else:
            result.append(char)

        if char in ".!?":
            capitalize_next = True

    return "".join(result)

sentence = input("print a sentence, and we'll clean it up")

result = clean_string(sentence)
print(result)
#raise NotImplementedError()
````
